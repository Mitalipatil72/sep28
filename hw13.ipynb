```python
a = 10
b = 3

print("Integer Arithmetic Operators:")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
print("Power:", a ** b)
print("Floor Division:", a // b)
```

```python
print("\nInteger Comparison Operators:")
print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)
```

```python
x = 10.5
y = 2.5

print("\nFloat Arithmetic Operators:")
print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Modulus:", x % y)
```

```python
print("\nFloat Comparison Operators:")
print("x == y:", x == y)
print("x != y:", x != y)
print("x > y:", x > y)
print("x < y:", x < y)
print("x >= y:", x >= y)
print("x <= y:", x <= y)
```

```python
s1 = "Hello"
s2 = "World"

print("\nString Operators:")
print("Concatenation:", s1 + " " + s2)
print("Repetition:", s1 * 2)

print("\nString Comparison Operators:")
print("s1 == s2:", s1 == s2)
print("s1 != s2:", s1 != s2)
print("s1 > s2:", s1 > s2)
print("s1 < s2:", s1 < s2)

```

```python
p = True
q = False

print("\nBoolean Comparison Operators:")
print("p == q:", p == q)
print("p != q:", p != q)
print("p > q:", p > q)
print("p < q:", p < q)
```
